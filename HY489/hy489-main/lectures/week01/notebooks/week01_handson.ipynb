### The same patient as a FHIR resource

From Week 3 onward we live in this format. For now, notice only that it is JSON,
that it is typed, and that fields can nest and repeat.


```python
import json

def to_fhir_patient(row):
    """One CSV row -> a minimal FHIR Patient resource."""
    return {
        "resourceType": "Patient",
        "id":        str(row["Id"]),
        "gender":    {"M": "male", "F": "female"}.get(row["GENDER"]),
        "birthDate": row["BIRTHDATE"].strftime("%Y-%m-%d"),
        "name":    [{"family": row["LAST"], "given": [row["FIRST"]]}],
        "address": [{"city": row["CITY"]}],
    }

print(json.dumps(to_fhir_patient(patients.iloc[0]), indent=2))
```

Compare that with the flat CSV row it came from. Same person, different shape.
